```python
print("="*120)
print()
print(                            "WEATHER TEMPERATURE ANALYSIS WITH NUMPY")
print()
print("___________________________________________________INRTRODUCTION___________________________________________________________")
print(f"""
Temperature data can provide useful information aout weather conditions across different locations and time periods. 
In this project, daily temperature were recorded by a weather monitoring station recorded for 5 days 
in 3 different locations; Lagos, Abuja and Ibadan.The temperatures are recorded as a raw numerical data but 
were analysed using NumPy. The project focuses on converting raw temperature data into NumPy arrays and aplying 
numerical operations to compare temperatures across the three locations.
The analysis also demonstrates how NUmPy can make the processing of numerical data more efficient and easier to manage.
""")

#import NumPy

import numpy as np

# Dataset of the three locations

lagos = [ 27.5,29.0, 30.5, 28.0, 31.0]
abuja = [ 25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [ 26.5, 28.5, 29.0, 27.5, 30.5]

# Convert the temeperature lists to arrays for each location

lagos_array = np.array(lagos)
abuja_array = np.array(abuja)
ibadan_array = np.array(ibadan)

print()
print("-"*70)
print("Converted Lagos Temeperature:", lagos_array)
print("Converted Abuja Temeperature:", abuja_array)
print("Converted Ibadan Temeperature:", ibadan_array)
print("-"*70)
print()

# Array Inspection

print("Lagos Data Type:", lagos_array.dtype)
print("Abuja Data Type:", abuja_array.dtype)
print("Ibadan Data Type:", ibadan_array.dtype)
print()

#Numerical Operations

print("-"*50)
print("_________________________________________________NUMERICAL OPERATIONS_________________________________________________________")
print()

#location that has the highest temperature recorded
print("Location with highest temeperature recorded :", np.maximum(lagos_array, abuja_array) )
print(f"""
Lagos recorded the highest temperature for the first three days, while Abuja recorded the highest for the last two days
""")
print()

# Calculate increase of recorded temperature by 1 degree celcius _

print("Increased Temperature for Lagos:", lagos_array + 1)
print("Increased Temperature for Abuja:", abuja_array + 1)
print("Increased Temperature for Ibadan:", ibadan_array +1)
print()

# Comparison betweeen Lagos temepratures and Ibadan temperatures

print("Compared Temepratures between Lagos and Abuja:", lagos_array - abuja_array)
print()

# Findings
print("________________________________________________________FINDINGS_______________________________________________________________")
print()
print(f"""
 1. Lagos recorded the highest temperatuee for the first three days, while Abuja recorded the highest temperature for the last two days.
 2. Lagos was warmer than Abuja during days 1 to 3, with temperature differences of 2.5◦C, 1.0◦C, and 1.0◦C respectively.
 3. Abuja recorded higher temperatures than Lagos on Day 4 and Day 5, exceeding Lagos by 2.0◦C and 1.0◦C respectively.
 4. NumPy successfully performed calculations across all temperature values simultaneously,such as incrasing each temperature by !◦C
    and calculating the differences between locations.
 5. NumPy makes numerical operations more efficient as the amount of data increases because it allows us to perform calculations on many
    values without repeating the process manually
""")
print()

# Conclusion

print("_______________________________________________________CONCLUSIONS________________________________________________________________")
print()
print(f"""
This minin project demonstrated the use of NumPy for analysing numerical temperature data from Lagos, Abuja, and Ibadan.
The temperature data were sucessfully converted from Python lists into NumPy arrays, inspected, and used for numeircal operations
such as addition, subtration and comparison of temperatures between locations. The analysi showed that Lagos recorded the highest
temperatures during the first three days, while Abuja recorded the highest temperatures during the last two days.
""")
print("="*120)
 
      
```
